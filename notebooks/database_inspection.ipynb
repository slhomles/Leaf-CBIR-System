# Kiểm tra dữ liệu trong Database
Notebook này kết nối đến cơ sở dữ liệu PostgreSQL và tải toàn bộ các bản ghi từ bảng `leaf_images` vào Pandas DataFrame để dễ dàng hiển thị dưới dạng bảng trực quan.

```python
import os
import sys
import pandas as pd

# Thêm thư mục gốc của dự án vào sys.path để import các module local
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from db.database import SessionLocal
from db.models import LeafImage
```

```python
def fetch_data_to_dataframe():
    # Tạo session kết nối CSDL
    db = SessionLocal()
    
    try:
        # Truy xuất tất cả các bản ghi
        records = db.query(LeafImage).all()
        
        if not records:
            print("Cơ sở dữ liệu đang trống.")
            return pd.DataFrame()
        
        # Chuyển đổi dữ liệu thành list of dictionaries
        data = []
        for r in records:
            data.append({
                'image_id': r.image_id,
                'file_name': r.file_name,
                'file_path': r.file_path,
                # Convert vectors to lists for display
                'shape_vector': list(r.shape_vector) if r.shape_vector is not None else None,
                'color_vector': list(r.color_vector) if r.color_vector is not None else None,
                'texture_vector': list(r.texture_vector) if r.texture_vector is not None else None,
                'venation_vector': list(r.venation_vector) if r.venation_vector is not None else None
            })
        
        # Tạo DataFrame
        df = pd.DataFrame(data)
        return df
    
    finally:
        db.close()

# Tải dữ liệu
df = fetch_data_to_dataframe()

# Hiển thị thông tin tổng quan
print(f"Tổng số bản ghi: {len(df)}")
```
